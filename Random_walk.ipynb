# Introduction
Random processes are very common in physics. When dealing with large numbers of molecules, such as in gases, we don’t describe the trajectory of every single particle ($10^{23}$ particles is too much to handle) but often use probabilities to determine the average behaviour. Here, we look at some illustrative examples of using random numbers in physical applications.

## Random walks
The random walk problem is a classic example which, despite its simplicity, describes very well the basic physical phenomenon of _diffusion_. Diffusion is a process where par- ticles which are initially concentrated in a region in space spread out with time. For example, a drop of ink in water will gradually spread to the whole volume of water, even if the water is perfectly still. This is because the individual molecules (both the ink and the water) (i) have random directions and speeds, which depend on the temperature, and (ii) also collide with each other. It turns out that these two ingredients can be modelled very easily.

## The method
We first look at a random walk in 1-D for a single particle. We start the particle off at the origin of the x-axis and assume that it moves unit distance in each time step (the molecular speed is thus 1 unit). At the end of each time step we set the direction to left or right with equal probability, mimicking the effect of collisions. You can easily write the python code, but for flexibility of later python coding, I show here how to define a function, called `walk(steps)`, which is passed a single number steps and returns a vector of that length, which contains all the positions of the random walk.

````python
def walk(steps):
    '''
    steps is the length of the 1-D random walk vector returned
    '''
    x = np.zeros(steps, int)  # initialise the array of integers
    x[0] = 0  # start at the origin
    for i in range(1, len(x)):
        rnd = random()  # rnd selected from [0,1)
        # step left or right randomly ...
        if (rnd<0.5):
            x[i] = x[i-1] + 1  # steps right one unit
        else:
            x[i] = x[i-1] - 1 # steps left one unit
    return x
````

IMPORTANT: Make sure you understand how this function is used, and how it works. The rest of this worksheet is based on using it. You can also use the vectorised version below.

````python
def walk(steps):
    '''
    steps is the length of the 1-D random walk vector returned
    Vectorised method.
    '''
    x = np.random.choice([-1, 1], size=steps) # random -1 or 1
    x = np.cumsum(x) # cumulative sum
    return x
````

## Prepare 3D Histogram Data

### Subtask:
Define histogram parameters, calculate histograms for each step, and prepare the data for 3D plotting.


**Reasoning**:
The subtask requires importing specific libraries, defining histogram parameters, calculating histograms for each step of the random walk, and structuring this data for 3D plotting. This comprehensive step will handle all these preparations in a single code block.



````python
from mpl_toolkits.mplot3d import Axes3D

# 1. Determine min and max position values across the entire particles array
min_pos = np.min(particles)
max_pos = np.max(particles)

# 2. Define the number of bins for the histograms
num_bins = 50

# 3. Create an array of bin edges and calculate bin centers
bin_edges = np.linspace(min_pos, max_pos, num_bins + 1)
bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2

# 4. Initialize a 2D NumPy array to store the frequency counts for each histogram
histogram_data = np.zeros((num_steps, num_bins), dtype=int)

# 5. Iterate through each step and compute the histogram
for s in range(num_steps):
    # Extract the positions of all particles at the current step
    positions_at_step_s = particles[s, :]
    # Compute the histogram for these positions
    counts, _ = np.histogram(positions_at_step_s, bins=bin_edges)
    # Store the calculated counts into the corresponding row of histogram_data
    histogram_data[s, :] = counts

# 6. Create two 2D meshgrid arrays, X and Y, for the 3D plot's axes
X_mesh, Y_mesh = np.meshgrid(bin_centers, np.arange(num_steps))

# 7. Assign the histogram_data array to a variable Z
Z_frequencies = histogram_data

print("Data for 3D histogram prepared: min_pos, max_pos, num_bins, bin_edges, bin_centers, histogram_data, X_mesh, Y_mesh, Z_frequencies are defined.")
````

## Create 3D Histogram Plot

### Subtask:
Generate a 3D surface plot showing the evolution of particle positions over time.


**Reasoning**:
The subtask is to create a 3D surface plot. I will use the prepared meshgrid data (X_mesh, Y_mesh, Z_frequencies) to generate the plot, setting appropriate labels and a title as requested.



````python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D # This import was already done in the previous code block, but is included for completeness within this step if run in isolation.

# Create a new figure and add a 3D subplot
fig = plt.figure(figsize=(14, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot the 3D surface
# X_mesh for particle positions, Y_mesh for step numbers, Z_frequencies for histogram frequencies
ax.plot_surface(X_mesh, Y_mesh, Z_frequencies, cmap='viridis', edgecolor='none')

# Set labels for the axes
ax.set_xlabel('Particle Position')
ax.set_ylabel('Step Number')
ax.set_zlabel('Frequency')

# Set the Z-axis limits to zoom in
ax.set_zlim(0, 150) # Adjust the upper limit as needed for desired zoom

# Add a title to the plot
ax.set_title('Evolution of Particle Position Distribution Over Time')

# Display the plot
plt.show()
````
